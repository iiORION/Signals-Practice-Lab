```python


import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 60)

# Two causal impulse responses
h_stable = 0.8 ** n
h_unstable = 1.1 ** n

plt.plot(n, h_stable, label="Stable System")
plt.plot(n, h_unstable, label="Unstable System")

plt.xlabel("Sample n")
plt.ylabel("Amplitude")
plt.title("LTI System Stability")
plt.grid(True)
plt.legend()
plt.show()

print("Stable partial sum:", np.sum(abs(h_stable)))
print("Unstable partial sum:", np.sum(abs(h_unstable)))
```
