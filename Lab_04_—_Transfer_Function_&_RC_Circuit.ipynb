```python

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

R = 2000
C = 200e-6

tau = R * C

# RC transfer function
system = signal.TransferFunction([1], [tau, 1])

# Step response
t = np.linspace(0, 1, 1000)
t, y = signal.step(system, T=t)

# Theoretical result
y_theory = 1 - np.exp(-t / tau)

plt.plot(t, y, label="Simulation")
plt.plot(t, y_theory, "--", label="Theory")
plt.xlabel("Time (s)")
plt.ylabel("Output Voltage (V)")
plt.title("RC Step Response")
plt.grid(True)
plt.legend()
plt.show()

print("Time constant =", tau, "seconds")
print("Cutoff frequency =", 1 / (2 * np.pi * tau), "Hz")

```
