```python

import numpy as np
import matplotlib.pyplot as plt

# Time axis
t = np.arange(-2, 2, 0.001)
dt = t[1] - t[0]

# Standard signals
rect = ((t >= -1) & (t < 1)).astype(float)
tri = np.maximum(1 - np.abs(t), 0)

# Calculate signal energy
E_rect = np.sum(rect**2) * dt
E_tri = np.sum(tri**2) * dt

print("Rect Energy =", round(E_rect, 3))
print("Tri Energy  =", round(E_tri, 3))

# Plot signals
plt.plot(t, rect, label="Rectangular")
plt.plot(t, tri, label="Triangular")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Lab 01 - Standard Signals")
plt.grid(True)
plt.legend()
plt.show()
```
