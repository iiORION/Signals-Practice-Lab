```python

import numpy as np
import matplotlib.pyplot as plt

# Create time values
t = np.linspace(-4, 4, 2001)

# Define the original signal
def x(time):
    return np.where(
        (time >= 0) & (time < 2),
        time,
        0
    )

# Signal transformations
original = x(t)
delayed = x(t - 1)
reversed_signal = x(-t)
compressed = x(2 * t)
amplified = 2 * x(t)

# Create four graphs
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

signals = [
    (delayed, "Time Shifting: x(t-1)"),
    (reversed_signal, "Time Reversal: x(-t)"),
    (compressed, "Time Compression: x(2t)"),
    (amplified, "Amplitude Scaling: 2x(t)")
]

for graph, (signal, title) in zip(ax.flat, signals):
    graph.plot(t, original, "--", label="Original")
    graph.plot(t, signal, label="Transformed")
    graph.set_title(title)
    graph.set_xlabel("Time (s)")
    graph.set_ylabel("Amplitude")
    graph.grid(True)
    graph.legend()

plt.tight_layout()
plt.show()
```
